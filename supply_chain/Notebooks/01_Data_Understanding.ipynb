# Analyse von Lieferkettenrisiken und Logistik

## Ziel des Projekts

Das Ziel dieses Projekts ist es, internationale Lieferdaten zu analysieren
und Faktoren zu untersuchen, die mit Lieferunterbrechungen und der
logistischen Leistung zusammenhängen.

Die Analyse konzentriert sich auf verschiedene Merkmale der Lieferungen,
Transportarten, Wetterbedingungen, geopolitische Risiken, die Zuverlässigkeit
der Transportunternehmen und die Lieferzeit.

## Forschungsfragen

1. Welche Faktoren hängen mit längeren Lieferzeiten zusammen?
2. Welche Faktoren stehen mit Lieferunterbrechungen in Zusammenhang?
3. Wie hängen Wetterbedingungen und geopolitische Risiken mit dem
   Risiko einer Lieferunterbrechung zusammen?
4. Kann eine einfache risikobasierte Methode dabei helfen, besonders
   gefährdete Lieferungen zu erkennen?

## Datensatz

Der Datensatz enthält 5.000 internationale Lieferungen aus den Jahren
2024–2025.

Er enthält Informationen über Lieferwege, Transportarten, Entfernung,
Gewicht, Kraftstoffpreise, geopolitische Risiken, Wetterbedingungen,
die Zuverlässigkeit der Transportunternehmen, Lieferzeiten und
Lieferunterbrechungen.

## Zielvariable

Die Variable `Disruption_Occurred` zeigt, ob eine Lieferung
unterbrochen wurde:

- `0` = Keine Unterbrechung
- `1` = Unterbrechung
