```python
import pandas as pd
```

```python
customers = pd.read_csv("train/df_Customers.csv")
order_items = pd.read_csv("train/df_OrderItems.csv")
orders = pd.read_csv("train/df_Orders.csv")
payments = pd.read_csv("train/df_Payments.csv")
products = pd.read_csv("train/df_Products.csv")
```

```python
print("Customers shape:", customers.shape)
print("Order Items shape:", order_items.shape)
print("Orders shape:", orders.shape)
print("Payments shape:", payments.shape)
print("Products shape:", products.shape)
```

```python
for name, df in [("Customers", customers), ("Order Items", order_items), ("Orders", orders), ("Payments", payments), ("Products", products)]:
    print(f"\n{name} DataFrame Info:")
    print(df.info())
    print(f"\n{name} DataFrame Description:")
    print(df.describe())
    print(f"\n{name} Missing Values:")
    print(df.isna().sum())
    print(f"\n{name} Duplicated Rows:")
    print(df.duplicated().sum())
```

```python
print("Total product rows: ", len(products))
print("Unique product IDs: ", products['product_id'].nunique())
print("Duplicated product IDs: ", products['product_id'].duplicated().sum())
```

```python
print("Customer")
print("Rows: ", len(customers))
print("Unique customer IDs: ", customers['customer_id'].nunique())
print("Duplicated customer IDs: ", customers['customer_id'].duplicated())

print("\nOrder Items")
print("Rows: ", len(order_items))
print("Unique order item IDs: ", order_items['order_item_id'].nunique())
print("Duplicated order item IDs: ", order_items['order_item_id'].duplicated())
```

```python
print("Product rows:", len(products))
print("Unique product IDs:", products["product_id"].nunique())
print("Duplicated product IDs:", products["product_id"].duplicated().sum())

print("Order rows:", len(orders))
print("Unique order IDs:", orders["order_id"].nunique())
print("Duplicated order IDs:", orders["order_id"].duplicated().sum())

print("Order Item rows:", len(order_items))
print("Unique order IDs in Order Items:", order_items["order_id"].nunique())
```

```python
print("Customer IDs in Order:", orders["customer_id"].nunique())
print("Customer IDs in Customers:", customers["customer_id"].nunique())

print(
    "Orders with customer IDs not found in Customers:", 
    (~orders["customer_id"].isin(customers["customer_id"])).sum()
)

print(
    "Order IDs in Order Items not found in Orders:",
    (~order_items["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Product IDs in Order Items not found in Products:",
    (~order_items["product_id"].isin(products["product_id"])).sum()
)
```

```python
date_columns = [
    "order_purchase_timestamp", 
    "order_approved_at", 
    "order_delivered_timestamp", 
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])

orders.dtypes
```

```python
orders["delivery_days"] = (orders["order_delivered_timestamp"] - orders["order_purchase_timestamp"]).dt.days
```
